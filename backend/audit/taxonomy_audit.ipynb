# Senseigigs Taxonomy Database Audit

This notebook generates a detailed, deterministic report from the **existing MySQL taxonomy database**. It is read-only: no records are inserted, updated, or deleted.

Before running it locally, start the database from the `backend` directory with `docker compose up -d --build`. The audit uses `MYSQL_*` environment variables when set and otherwise connects to `127.0.0.1:3307`.

```python
from pathlib import Path
import sys

# Support launching Jupyter from either backend/ or backend/audit/.
working_directory = Path.cwd().resolve()
audit_directory = working_directory / 'audit' if (working_directory / 'audit').is_dir() else working_directory
if str(audit_directory) not in sys.path:
    sys.path.insert(0, str(audit_directory))

from taxonomy_audit import database_config, generate_report

config = database_config()
print(f"Audit target: {config['host']}:{config['port']}/{config['database']}")
```

```python
from IPython.display import Markdown, display

try:
    report = generate_report(config)
except Exception as exc:
    raise RuntimeError(
        f"Could not audit {config['database']} at {config['host']}:{config['port']}. "
        "Start Docker with `docker compose up -d --build` from the backend directory, "
        "then rerun this cell."
    ) from exc

display(Markdown(report))
```

## Optional export

Run the next cell to save the exact rendered source as a portable Markdown report beside this notebook.

```python
output_path = audit_directory / 'taxonomy_audit_report.md'
output_path.write_text(report, encoding='utf-8')
print(f'Saved report to {output_path}')
```
